```python
# Execute this cell
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt
%matplotlib inline
%config InlineBackend.figure_format='retina'
```

```python
M = 10000
N = 20
volume = np.array([])
for j in range (N):

    x = np.zeros((j + 1 , M))
    r = 0
    for i in range (j + 1):
        x[i , :] = stats.uniform(0,1).rvs(M)
        r += x[i , :]**2 # equation for radius of cirle in x,y

    volume = np.append(volume, np.size(x[0][r < 1])/np.size(x[0]))

#plt.plot(x[0], x[1], '.', markersize=3, c='blue') 

#plt.plot(x[0][r < 1], x[1][r < 1], '.', markersize=3, c='red') 
#plt.show
x = np.linspace(1 , N , N)
plt.plot(x, volume )
volume
```
