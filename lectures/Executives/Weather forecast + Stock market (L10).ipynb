```python
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt
%matplotlib inline
%config InlineBackend.figure_format='retina'
```

```python
#we'll use the '-1' to identify cloudy weather and '1' to identify clear one

N = 40000 #number of iterations

W = np.zeros(N) #array of weather
W[0] = -1  #starting off with cloudy weather
clear = np.array([]) 
cloudy = np.array([]) 

for i in range (N-1):
    check = stats.uniform(0,1).rvs(1)
    if (W[i] == -1):
        if(check > 0.5):
            W[i + 1] = -1
        else:
            W[i + 1] = 1
    if (W[i] == 1):
        if(check > 0.9):
            W[i + 1] = -1
        else:
            W[i + 1] = 1
    clear = np.append(clear , np.size( np.where ( W == 1 ))/ (i+2))
    cloudy = np.append(cloudy , np.size( np.where ( W == -1 ))/ (i+2))


print( cloudy[-1] , clear[-1])
```

```python
#plotto i risultati:
xgrid = np.linspace(1, N , N-1)
plt.plot(xgrid,clear , color = 'orange')
plt.plot(xgrid,cloudy , color = 'blue')
```

```python
burn = N/2
burn = int(burn)
#data shown in the histogram after thermalization
plt.hist(clear[burn:] , bins = 'auto' , density = True ,  alpha = 0.5)

#computing summary statistics
mean = np.mean(clear[burn:])
var = np.var(clear[burn:])

#show mean on the graph
mean_rep = np.ones(10000)
y = mean_rep * mean
x = np.linspace ( 0 , 600 , 10000)

plt.plot(y , x , color = 'red')
plt.show()
```

STOCK MARKET

```python
#we'll use the '1' to identify BULL MARKET , '2' to identify BEAR MARKET , '3' to indentify STAGNANT MARKET

N = 20000 #number of iterations

W = np.zeros(N) #array of markets
W[0] = 1  #starting off with BULL MARKET
BULL = np.array([]) 
BEAR = np.array([]) 
STAGNANT = np.array([])

for i in range (N-1):
    check = stats.uniform(0,1).rvs(1)
    if (W[i] == 1):
        if(check < 0.025):
            W[i + 1] = 3
        elif(check > 0.025 and check < 0.1 ):
            W[i + 1] = 2
        else:
            W[i + 1] = 1
    if (W[i] == 2):
        if(check < 0.05):
            W[i + 1] = 3
        elif(check > 0.05 and check < 0.2 ):
            W[i + 1] = 1
        else:
            W[i + 1] = 2
    if (W[i] == 3):
        if(check < 0.25):
            W[i + 1] = 1
        elif(check > 0.25 and check < 0.5 ):
            W[i + 1] = 2
        else:
            W[i + 1] = 3
    BULL = np.append(BULL , np.size( np.where ( W == 1 ))/ (i+2))
    BEAR = np.append(BEAR , np.size( np.where ( W == 2 ))/ (i+2))
    STAGNANT = np.append(STAGNANT , np.size( np.where ( W == 3 ))/ (i+2))


```

```python
#plotto i risultati:
xgrid = np.linspace(1, N , N-1)
plt.plot(xgrid,BULL , color = 'orange')
plt.plot(xgrid,BEAR , color = 'blue')
plt.plot(xgrid,STAGNANT,color = 'green')
plt.show()
```
